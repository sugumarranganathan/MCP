Create your first MCP tool

```python
from mcp.server import MCPServer

mcp = MCPServer("EV Data Server")


@mcp.tool()
def get_ev_charging_data() -> str:
    """Return today's EV charging data."""

    return """
    Date: 08-08-2026
    Total EV Charging Sessions: 125
    Total Energy Consumed: 1840 kWh
    Peak Hour: 7 PM
    Average Session Energy: 14.72 kWh
    """


print("MCP Server created successfully!")
```

You have now created an MCP server with one tool:

get_ev_charging_data()

create an MCP client

```python
from mcp import Client
```

```python
async with Client(mcp) as client:
    tools = await client.list_tools()
    print(tools)
```

Call the MCP tool

```python
async with Client(mcp) as client:
    result = await client.call_tool(
        "get_ev_charging_data",
        {}
    )

    print(result)
```

What happened

You asked the MCP client to execute:

get_ev_charging_data()

MCP routed the request:

MCP Client
    ↓
    ↓ MCP
    ↓
MCP Server
    ↓
get_ev_charging_data()
    ↓
EV charging data
    ↓
MCP Client

And you received:

| Data                   |     Result |
| ---------------------- | ---------: |
| Date                   | 08-08-2026 |
| EV Charging Sessions   |        125 |
| Energy Consumed        |  1,840 kWh |
| Peak Hour              |       7 PM |
| Average Session Energy |  14.72 kWh |

What is_error=False means

This is especially important:

is_error=False

means:

✅ The MCP tool executed successfully without an error.

And:

result_type='complete'

means the MCP response completed successfully.
